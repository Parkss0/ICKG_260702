```python
import networkx as nx
import plotly.graph_objects as go

def visualize_gene_to_pathway(G, genes, pathway, max_path_len=6):
    paths = []

    for gene in genes:
        if gene not in G:
            print(f"{gene} not found in graph")
            continue

        if pathway not in G:
            print(f"{pathway} not found in graph")
            continue

        try:
            path = nx.shortest_path(G, source=gene, target=pathway)
            if len(path) <= max_path_len:
                paths.append(path)
            else:
                print(f"{gene} to {pathway}: path too long, length={len(path)}")
        except nx.NetworkXNoPath:
            print(f"No path from {gene} to {pathway}")

    if not paths:
        print("No valid paths found")
        return None

    # path에 포함된 node와 edge만 subgraph로 추출
    nodes = set()
    edges = []

    for path in paths:
        nodes.update(path)
        for i in range(len(path) - 1):
            edges.append((path[i], path[i + 1]))

    subG = G.subgraph(nodes).copy()

    # layout
    pos = nx.spring_layout(subG, seed=42, k=1.2)

    edge_x = []
    edge_y = []
    edge_text = []

    for u, v in edges:
        x0, y0 = pos[u]
        x1, y1 = pos[v]

        edge_x += [x0, x1, None]
        edge_y += [y0, y1, None]

        rel = G[u][v].get("relationship", "")
        sources = G[u][v].get("sources", [])
        pmids = ", ".join([str(s).replace("[", "").replace("]", "").replace(".t", "") for s in sources[:5]])

        edge_text += [
            f"{u} → {v}<br>Relationship: {rel}<br>PMIDs: {pmids}",
            f"{u} → {v}<br>Relationship: {rel}<br>PMIDs: {pmids}",
            None
        ]

    edge_trace = go.Scatter(
        x=edge_x,
        y=edge_y,
        line=dict(width=2),
        hoverinfo="text",
        text=edge_text,
        mode="lines"
    )

    node_x = []
    node_y = []
    node_text = []
    node_size = []

    for node in subG.nodes():
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)

        node_type = G.nodes[node].get("node_type", "")
        node_text.append(f"{node}<br>Type: {node_type}")

        if node in genes:
            node_size.append(28)
        elif node == pathway:
            node_size.append(36)
        else:
            node_size.append(20)

    node_trace = go.Scatter(
        x=node_x,
        y=node_y,
        mode="markers+text",
        text=list(subG.nodes()),
        hovertext=node_text,
        hoverinfo="text",
        textposition="top center",
        marker=dict(
            size=node_size,
            line=dict(width=2)
        )
    )

    fig = go.Figure(data=[edge_trace, node_trace])

    fig.update_layout(
        title=f"ICKG path network: {', '.join(genes)} → {pathway}",
        showlegend=False,
        hovermode="closest",
        width=900,
        height=650,
        margin=dict(l=20, r=20, t=60, b=20),
        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
    )

    fig.show()
    return fig
```

```python
# ============================================
# 발표용 KG 시각화 (핵심 엣지만)
# ============================================
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

def visualize_ickg(cell_type="T", top_n=40):
    """
    cell_type: "T", "B", "NK" 중 선택
    top_n: weight 상위 몇 개 엣지를 그릴지
    """
    # 1. 데이터 로드
    edges = pd.read_csv(f"portal/data/{cell_type}_relationships_pmid_gene_normalized.csv")
    nodes = pd.read_csv(f"portal/data/{cell_type}_nodes_pmid_gene_normalized.csv")

    # 2. node_type 매핑 (색깔용)
    type_map = dict(zip(nodes['name'], nodes['node_type']))

    # 3. weight 상위 top_n개만 추출 (핵심 관계)
    top_edges = edges.nlargest(top_n, 'weight:int')

    print(f"=== {cell_type} cell KG ===")
    print(f"전체 엣지: {len(edges):,} → 상위 {top_n}개만 시각화")
    print(f"weight 범위: {top_edges['weight:int'].min()} ~ {top_edges['weight:int'].max()}\n")

    # 4. 그래프 구성
    G = nx.DiGraph()
    for _, row in top_edges.iterrows():
        G.add_edge(
            row[':START_ID'], row[':END_ID'],
            relationship=row['relationship'],
            weight=row['weight:int']
        )

    # 5. 노드 색 (종류별)
    color_by_type = {
        "gene": "#4ECDC4",      # teal
        "disease": "#FF6B6B",   # red
        "cell type": "#FFE66D", # yellow
        "unknown": "#C7CEEA",   # light purple
    }
    node_colors = []
    for n in G.nodes():
        ntype = type_map.get(n, "unknown")
        node_colors.append(color_by_type.get(ntype, "#CCCCCC"))

    # 6. 노드 크기 (연결 많을수록 크게)
    node_sizes = [300 + G.degree(n) * 150 for n in G.nodes()]

    # 7. 그리기
    plt.figure(figsize=(16, 12))
    pos = nx.spring_layout(G, k=1.5, iterations=80, seed=42)

    nx.draw_networkx_nodes(G, pos, node_color=node_colors,
                           node_size=node_sizes, alpha=0.9)
    nx.draw_networkx_labels(G, pos, font_size=8, font_weight="bold")

    # activate=초록, inhibit=빨강
    act = [(u,v) for u,v,d in G.edges(data=True) if d["relationship"]=="activate"]
    inh = [(u,v) for u,v,d in G.edges(data=True) if d["relationship"]=="inhibit"]

    # 엣지 두께 = weight 비례
    act_w = [G[u][v]["weight"]*0.3 for u,v in act]
    inh_w = [G[u][v]["weight"]*0.3 for u,v in inh]

    nx.draw_networkx_edges(G, pos, edgelist=act, edge_color="#2ECC71",
                           width=act_w, arrowsize=12, arrowstyle="-|>",
                           alpha=0.6, connectionstyle="arc3,rad=0.1")
    nx.draw_networkx_edges(G, pos, edgelist=inh, edge_color="#E74C3C",
                           width=inh_w, arrowsize=12, arrowstyle="-|>",
                           alpha=0.6, style="dashed", connectionstyle="arc3,rad=0.1")

    # 범례
    legend = [
        Line2D([0],[0], color="#2ECC71", lw=2, label="Activate"),
        Line2D([0],[0], color="#E74C3C", lw=2, ls="--", label="Inhibit"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#4ECDC4", markersize=12, label="Gene"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#FF6B6B", markersize=12, label="Disease"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#FFE66D", markersize=12, label="Cell type"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#C7CEEA", markersize=12, label="Other"),
    ]
    plt.legend(handles=legend, loc="upper left", fontsize=11)

    plt.title(f"Immune Cell Knowledge Graph — {cell_type} cell\n"
              f"(Top {top_n} edges by literature support, from He et al. ICKG)",
              fontsize=15, fontweight="bold")
    plt.axis("off")
    plt.tight_layout()
    plt.savefig(f"ickg_{cell_type}_top{top_n}.png", dpi=200, bbox_inches="tight")
    plt.show()
    print(f"저장됨: ickg_{cell_type}_top{top_n}.png")


# ----- 실행: T세포와 B세포 (네 캐스케이드에 맞게) -----
visualize_ickg(cell_type="T", top_n=40)
visualize_ickg(cell_type="B", top_n=40)
```

```python
# ============================================
# 논문식 원형 KG 시각화 (He et al. Figure 2 스타일)
# ============================================
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

def visualize_ickg_circular(cell_type="B", top_n=45):
    edges = pd.read_csv(f"portal/data/{cell_type}_relationships_pmid_gene_normalized.csv")
    nodes = pd.read_csv(f"portal/data/{cell_type}_nodes_pmid_gene_normalized.csv")
    type_map = dict(zip(nodes['name'], nodes['node_type']))

    top_edges = edges.nlargest(top_n, 'weight:int')

    G = nx.DiGraph()
    for _, row in top_edges.iterrows():
        G.add_edge(row[':START_ID'], row[':END_ID'],
                   relationship=row['relationship'], weight=row['weight:int'])

    # --- 원형 배치: 연결 많은 노드(허브)를 기준으로 정렬 ---
    degree = dict(G.degree())
    ordered = sorted(G.nodes(), key=lambda n: degree[n], reverse=True)
    pos = nx.circular_layout(ordered)  # 원 둘레에 배치

    # 색 (논문식 톤 다운 팔레트)
    color_by_type = {
        "gene": "#5B9BD5", "disease": "#C0504D",
        "cell type": "#9BBB59", "unknown": "#B0A8B9",
    }
    node_colors = [color_by_type.get(type_map.get(n, "unknown"), "#B0A8B9") for n in G.nodes()]
    node_sizes  = [200 + degree[n]*120 for n in G.nodes()]

    fig, ax = plt.subplots(figsize=(15, 15))

    # 노드
    nx.draw_networkx_nodes(G, pos, node_color=node_colors,
                           node_size=node_sizes, alpha=0.95,
                           edgecolors="white", linewidths=1.5, ax=ax)

    # 라벨을 원 바깥으로 밀어서 겹침 방지
    label_pos = {n: (x*1.12, y*1.12) for n, (x,y) in pos.items()}
    for n,(x,y) in label_pos.items():
        angle = np.degrees(np.arctan2(y, x))
        ha = "left" if x >= 0 else "right"
        ax.text(x, y, n, fontsize=8, fontweight="bold",
                ha=ha, va="center", rotation=0)

    # 엣지: 안쪽으로 휘는 호, activate=청록 / inhibit=적색
    act = [(u,v) for u,v,d in G.edges(data=True) if d["relationship"]=="activate"]
    inh = [(u,v) for u,v,d in G.edges(data=True) if d["relationship"]=="inhibit"]

    maxw = top_edges['weight:int'].max()
    nx.draw_networkx_edges(G, pos, edgelist=act, edge_color="#4C9A8E",
        width=[G[u][v]["weight"]/maxw*4 for u,v in act],
        alpha=0.5, arrowsize=8, arrowstyle="-|>",
        connectionstyle="arc3,rad=0.25", ax=ax)
    nx.draw_networkx_edges(G, pos, edgelist=inh, edge_color="#C0504D",
        width=[G[u][v]["weight"]/maxw*4 for u,v in inh],
        alpha=0.6, arrowsize=8, arrowstyle="-|>", style="dashed",
        connectionstyle="arc3,rad=0.25", ax=ax)

    legend = [
        Line2D([0],[0], color="#4C9A8E", lw=2, label="Activate"),
        Line2D([0],[0], color="#C0504D", lw=2, ls="--", label="Inhibit"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#5B9BD5", markersize=11, label="Gene"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#C0504D", markersize=11, label="Disease"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#9BBB59", markersize=11, label="Cell type"),
        Line2D([0],[0], marker="o", color="w", markerfacecolor="#B0A8B9", markersize=11, label="Other"),
    ]
    ax.legend(handles=legend, loc="upper left", fontsize=11, frameon=True)
    ax.set_title(f"Immune Cell Knowledge Graph — {cell_type} cell\n"
                 f"Top {top_n} edges by literature support (He et al. ICKG)",
                 fontsize=15, fontweight="bold", pad=20)
    ax.set_xlim(-1.3, 1.3); ax.set_ylim(-1.3, 1.3)
    ax.axis("off")
    plt.tight_layout()
    plt.savefig(f"ickg_circular_{cell_type}.png", dpi=200, bbox_inches="tight")
    plt.show()
    print(f"저장됨: ickg_circular_{cell_type}.png")


visualize_ickg_circular("B", top_n=45)
visualize_ickg_circular("T", top_n=45)
```

```python
# ============================================
# 0️⃣ 필요한 라이브러리 (이미 설치됨)
# ============================================
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import numpy as np
import os

data_dir = "portal/data"

print("=" * 90)
print("STEP 1: NK/B/T 엣지 파일 로드")
print("=" * 90)

cell_types = {
    "NK": f"{data_dir}/NK_relationships_pmid_gene_normalized.csv",
    "B": f"{data_dir}/B_relationships_pmid_gene_normalized.csv",
    "T": f"{data_dir}/T_relationships_pmid_gene_normalized.csv"
}

edges_all = {}
for cell_name, fpath in cell_types.items():
    df = pd.read_csv(fpath)
    edges_all[cell_name] = df
    print(f"✅ {cell_name}: {len(df):,} edges loaded")

print("\n" + "=" * 90)
print("STEP 2: 바이러스 키워드로 필터링")
print("=" * 90)

keywords = ["sars", "cov", "covid", "viral", "virus", "spike", "antibody",
            "neutraliz", "escape", "immune escape", "variant", "mutation"]

print(f"\n📍 필터링 키워드: {keywords}\n")

virus_edges_by_cell = {}
for cell_name, edges_df in edges_all.items():
    mask = (
        edges_df[':START_ID'].str.lower().str.contains('|'.join(keywords), na=False) |
        edges_df[':END_ID'].str.lower().str.contains('|'.join(keywords), na=False)
    )
    virus_edges = edges_df[mask].copy()
    virus_edges_by_cell[cell_name] = virus_edges
    pct = (len(virus_edges) / len(edges_df)) * 100
    print(f"   {cell_name}: {len(virus_edges):5d} / {len(edges_df):6,d}  ({pct:5.2f}%)")

print("\n" + "=" * 90)
print("STEP 3: 노드 타입 분류")
print("=" * 90)

nodes_by_cell = {}
node_info = {}

for cell_name, npath in [("NK", f"{data_dir}/NK_nodes_pmid_gene_normalized.csv"),
                         ("B", f"{data_dir}/B_nodes_pmid_gene_normalized.csv"),
                         ("T", f"{data_dir}/T_nodes_pmid_gene_normalized.csv")]:
    nodes_df = pd.read_csv(npath)
    nodes_by_cell[cell_name] = nodes_df

    for idx, row in nodes_df.iterrows():
        node_id = row['nodeId:ID']
        node_type = row['node_type']
        if node_id not in node_info:
            node_info[node_id] = {
                "type": node_type,
                "cell_sources": [cell_name]
            }
        else:
            if cell_name not in node_info[node_id]["cell_sources"]:
                node_info[node_id]["cell_sources"].append(cell_name)

    print(f"✅ {cell_name}: {len(nodes_df):,} nodes loaded")

print(f"\n📊 총 unique nodes: {len(node_info):,}")

print("\n" + "=" * 90)
print("STEP 4: 바이러스 엣지 상세 분석")
print("=" * 90)

all_virus_edges = pd.concat(
    [df.assign(source_cell=cell) for cell, df in virus_edges_by_cell.items()],
    ignore_index=True
)

print(f"\n📌 총 바이러스 엣지: {len(all_virus_edges):,} (중복 포함)")
print(f"\n   Relationship type 분포:")
rel_dist = all_virus_edges['relationship'].value_counts()
for rel, count in rel_dist.items():
    print(f"      {rel:20s}: {count:5d}")

print(f"\n   Weight 분포:")
print(f"      Min: {all_virus_edges['weight:int'].min()}")
print(f"      Max: {all_virus_edges['weight:int'].max()}")
print(f"      Mean: {all_virus_edges['weight:int'].mean():.2f}")
print(f"      Median: {all_virus_edges['weight:int'].median():.2f}")

print("\n" + "=" * 90)
print("STEP 5: NetworkX 그래프 생성")
print("=" * 90)

G = nx.DiGraph()

for idx, row in all_virus_edges.iterrows():
    start = row[':START_ID']
    end = row[':END_ID']
    rel = row['relationship']
    weight = row['weight:int']
    cell_src = row['source_cell']

    if start not in G:
        G.add_node(start)
    if end not in G:
        G.add_node(end)

    if G.has_edge(start, end):
        G[start][end]['weight'] += weight
        G[start][end]['relationships'].append(rel)
        G[start][end]['cell_sources'].append(cell_src)
    else:
        G.add_edge(start, end,
                  weight=weight,
                  relationship=rel,
                  relationships=[rel],
                  cell_sources=[cell_src])

print(f"\n✅ 그래프 생성 완료")
print(f"   Nodes: {G.number_of_nodes()}")
print(f"   Edges: {G.number_of_edges()}")

for node in G.nodes():
    if node in node_info:
        G.nodes[node]['node_type'] = node_info[node]['type']
        G.nodes[node]['cell_sources'] = node_info[node]['cell_sources']
    else:
        G.nodes[node]['node_type'] = 'unknown'
        G.nodes[node]['cell_sources'] = []

print("\n" + "=" * 90)
print("STEP 6: 노드 카테고리 분류")
print("=" * 90)

node_categories = {
    'virus': [],
    'gene': [],
    'cell': [],
    'concept': []
}

virus_keywords = ["covid", "sars", "cov", "virus", "spike"]

for node in G.nodes():
    node_lower = node.lower()

    if any(kw in node_lower for kw in virus_keywords):
        node_categories['virus'].append(node)
        G.nodes[node]['category'] = 'virus'
    elif node in node_info:
        ntype = node_info[node]['type']
        if ntype == 'gene':
            node_categories['gene'].append(node)
            G.nodes[node]['category'] = 'gene'
        elif ntype in ['disease', 'cell', 'unknown']:
            if 'cell' in node_lower or 'lymph' in node_lower:
                node_categories['cell'].append(node)
                G.nodes[node]['category'] = 'cell'
            else:
                node_categories['concept'].append(node)
                G.nodes[node]['category'] = 'concept'
        else:
            node_categories['concept'].append(node)
            G.nodes[node]['category'] = 'concept'
    else:
        if 'cell' in node_lower or 'lymph' in node_lower:
            node_categories['cell'].append(node)
            G.nodes[node]['category'] = 'cell'
        else:
            node_categories['concept'].append(node)
            G.nodes[node]['category'] = 'concept'

print(f"\n📊 노드 분류:")
for cat, nodes in node_categories.items():
    print(f"   {cat:10s}: {len(nodes):5d}")

print("\n   Virus nodes:")
for node in node_categories['virus']:
    print(f"      - {node}")

print("\n" + "=" * 90)
print("STEP 7: 시각화 준비")
print("=" * 90)

print("\n⏳ Spring layout 계산 중...")
pos = nx.spring_layout(G, k=2, iterations=50, seed=42)

node_colors = []
node_sizes = []
color_map = {
    'virus': '#FF4444',
    'gene': '#888888',
    'cell': '#4488FF',
    'concept': '#CCCCCC'
}

size_map = {
    'virus': 1000,
    'cell': 600,
    'gene': 300,
    'concept': 200
}

for node in G.nodes():
    cat = G.nodes[node].get('category', 'concept')
    node_colors.append(color_map.get(cat, '#CCCCCC'))
    node_sizes.append(size_map.get(cat, 200))

edge_colors = []
edge_widths = []
edge_styles = []

for start, end in G.edges():
    rel = G[start][end]['relationship']
    weight = G[start][end]['weight']

    if 'activate' in rel.lower():
        edge_colors.append('#00AA00')
    elif 'inhibit' in rel.lower():
        edge_colors.append('#AA0000')
    else:
        edge_colors.append('#0066FF')

    edge_widths.append(0.5 + min(weight, 5) * 0.3)

    if 'activate' in rel.lower():
        edge_styles.append('solid')
    elif 'inhibit' in rel.lower():
        edge_styles.append('dashed')
    else:
        edge_styles.append('dotted')

print(f"✅ 시각화 속성 설정 완료")

print("\n" + "=" * 90)
print("STEP 8: 그래프 그리기")
print("=" * 90)

fig, ax = plt.subplots(figsize=(20, 16), dpi=100)

for style in ['solid', 'dashed', 'dotted']:
    edges_of_style = [(start, end) for start, end in G.edges()
                      if edge_styles[list(G.edges()).index((start, end))] == style]
    widths_of_style = [edge_widths[list(G.edges()).index((start, end))]
                       for start, end in edges_of_style]
    colors_of_style = [edge_colors[list(G.edges()).index((start, end))]
                       for start, end in edges_of_style]

    nx.draw_networkx_edges(G, pos,
                          edgelist=edges_of_style,
                          width=widths_of_style,
                          edge_color=colors_of_style,
                          style=style,
                          ax=ax,
                          alpha=0.6,
                          arrows=True,
                          arrowsize=15,
                          connectionstyle="arc3,rad=0.1")

nx.draw_networkx_nodes(G, pos,
                       node_color=node_colors,
                       node_size=node_sizes,
                       ax=ax,
                       alpha=0.9,
                       edgecolors='black',
                       linewidths=1.5)

labels = {}
for node in G.nodes():
    cat = G.nodes[node].get('category', 'concept')
    if cat == 'virus' or G.degree(node) > 5:
        labels[node] = node

nx.draw_networkx_labels(G, pos,
                        labels=labels,
                        font_size=9,
                        font_weight='bold',
                        ax=ax)

legend_elements = [
    Patch(facecolor='#FF4444', edgecolor='black', label='Virus'),
    Patch(facecolor='#4488FF', edgecolor='black', label='Cell'),
    Patch(facecolor='#888888', edgecolor='black', label='Gene'),
    Patch(facecolor='#CCCCCC', edgecolor='black', label='Concept'),
    plt.Line2D([0], [0], color='#00AA00', linewidth=2, label='Activate'),
    plt.Line2D([0], [0], color='#AA0000', linewidth=2, label='Inhibit'),
    plt.Line2D([0], [0], color='#0066FF', linewidth=2, linestyle='dotted', label='Other'),
]
ax.legend(handles=legend_elements, loc='upper left', fontsize=12, framealpha=0.95)

ax.set_title(f"Viral Immune Escape Network (NK/B/T)\n"
             f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}",
             fontsize=16, fontweight='bold', pad=20)
ax.axis('off')
plt.tight_layout()
plt.savefig('viral_network_graph.png', dpi=150, bbox_inches='tight')
print(f"\n✅ 그래프 저장: viral_network_graph.png")
plt.show()

print("\n" + "=" * 90)
print("STEP 9: 네트워크 통계")
print("=" * 90)

print(f"\n📊 그래프 통계:")
print(f"   Nodes: {G.number_of_nodes()}")
print(f"   Edges: {G.number_of_edges()}")
print(f"   Density: {nx.density(G):.4f}")
print(f"   Average degree: {sum(dict(G.degree()).values()) / G.number_of_nodes():.2f}")

print(f"\n🔍 Degree centrality 상위 10:")
degree_cent = nx.degree_centrality(G)
for node, cent in sorted(degree_cent.items(), key=lambda x: x[1], reverse=True)[:10]:
    print(f"   {node:30s}: {cent:.4f} (degree: {G.degree(node)})")

print(f"\n📝 Relationship 통계:")
rel_summary = all_virus_edges.groupby('relationship').agg({
    'weight:int': ['count', 'sum', 'mean']
}).round(2)
print(rel_summary)

print("\n" + "=" * 90)
print("✅ COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# ICKG 통합 로딩 + SARS-CoV-2 직접 연결 엣지 정밀 감사
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re
import ast

DATA_DIR = Path("portal/data")


# ------------------------------------------------------------
# 1. 모든 edge / node 파일 탐색
# ------------------------------------------------------------

edge_files = sorted(
    DATA_DIR.glob("*_relationships_pmid_gene_normalized.csv")
)

node_files = sorted(
    DATA_DIR.glob("*_nodes_pmid_gene_normalized.csv")
)

print("=== 발견된 관계 파일 ===")
for path in edge_files:
    print(" ", path.name)

print("\n=== 발견된 노드 파일 ===")
for path in node_files:
    print(" ", path.name)


# ------------------------------------------------------------
# 2. 관계 파일 통합
# ------------------------------------------------------------

edge_tables = []

for path in edge_files:

    cell_context = path.name.split("_relationships")[0]

    df = pd.read_csv(path)

    required_columns = {
        ":START_ID",
        ":END_ID",
        "relationship",
        "weight:int",
        "sources"
    }

    missing = required_columns - set(df.columns)

    if missing:
        raise ValueError(
            f"{path.name}에 필요한 컬럼이 없습니다: {missing}"
        )

    df = df.rename(
        columns={
            ":START_ID": "source",
            ":END_ID": "target",
            "weight:int": "weight",
            "file_count:int": "file_count"
        }
    )

    df["cell_context"] = cell_context
    df["source_file"] = path.name

    edge_tables.append(df)

edges = pd.concat(edge_tables, ignore_index=True)

print(f"\n통합 엣지 수: {len(edges):,}")


# ------------------------------------------------------------
# 3. 노드 파일 통합
# ------------------------------------------------------------

node_tables = []

for path in node_files:

    cell_context = path.name.split("_nodes")[0]

    df = pd.read_csv(path)

    df = df.rename(
        columns={
            "nodeId:ID": "node_id",
            ":LABEL": "label"
        }
    )

    df["cell_context"] = cell_context
    df["source_file"] = path.name

    node_tables.append(df)

nodes = pd.concat(node_tables, ignore_index=True)

print(f"통합 노드 행 수: {len(nodes):,}")


# ------------------------------------------------------------
# 4. 문자열 표준화
# ------------------------------------------------------------

def normalize_term(value):

    if pd.isna(value):
        return ""

    value = str(value).strip().lower()

    # 여러 종류의 dash를 일반 하이픈으로 통일
    value = re.sub(r"[‐-‒–—_]", "-", value)

    # 연속 공백 제거
    value = re.sub(r"\s+", " ", value)

    return value


edges["source_norm"] = edges["source"].map(normalize_term)
edges["target_norm"] = edges["target"].map(normalize_term)

nodes["node_norm"] = nodes["node_id"].map(normalize_term)
nodes["name_norm"] = nodes["name"].map(normalize_term)


# ------------------------------------------------------------
# 5. PMID 목록 파싱
# ------------------------------------------------------------

def parse_pmids(value):

    if pd.isna(value):
        return []

    try:
        parsed = ast.literal_eval(str(value))

        if not isinstance(parsed, list):
            parsed = [parsed]

    except (ValueError, SyntaxError):
        parsed = re.findall(r"\d{7,9}", str(value))

    pmids = []

    for item in parsed:
        match = re.search(r"\d{7,9}", str(item))

        if match:
            pmids.append(match.group())

    return sorted(set(pmids))


edges["pmids"] = edges["sources"].map(parse_pmids)
edges["pmid_count_parsed"] = edges["pmids"].map(len)


# ------------------------------------------------------------
# 6. 검색 패턴 정의
# ------------------------------------------------------------

# SARS-CoV-2에 직접 해당하는 강한 seed
STRICT_PATTERNS = [
    r"\bsars[\s-]*cov[\s-]*2\b",
    r"\bcovid[\s-]*19\b",
    r"\b2019[\s-]*ncov\b",
    r"\bsevere acute respiratory syndrome coronavirus 2\b"
]

# 일반적인 바이러스 관련 용어
GENERIC_VIRUS_PATTERNS = [
    r"\bcoronavirus(?:es)?\b",
    r"\bvirus(?:es)?\b",
    r"\bviral\b",
    r"\bspike protein\b",
    r"\bspike glycoprotein\b"
]

# 이것만으로는 바이러스 관계라고 볼 수 없는 용어
IMMUNE_GENERAL_PATTERNS = [
    r"\bantibod",
    r"\bneutraliz",
    r"\bimmune escape\b",
    r"\bimmune evasion\b"
]


def regex_mask(series, patterns):

    pattern = "|".join(
        f"(?:{item})" for item in patterns
    )

    return series.fillna("").str.contains(
        pattern,
        regex=True,
        case=False
    )


# ------------------------------------------------------------
# 7. SARS-CoV-2 seed 노드 확인
# ------------------------------------------------------------

strict_node_mask = (
    regex_mask(nodes["node_norm"], STRICT_PATTERNS)
    | regex_mask(nodes["name_norm"], STRICT_PATTERNS)
)

strict_nodes = (
    nodes.loc[
        strict_node_mask,
        ["cell_context", "node_id", "name", "node_type"]
    ]
    .drop_duplicates()
    .sort_values(["cell_context", "node_id"])
)

print("\n=== SARS-CoV-2 / COVID-19 seed 노드 ===")
print(strict_nodes.to_string(index=False))


seed_names = set(
    strict_nodes["node_id"].map(normalize_term)
)


# 혹시 node 파일에 누락된 seed가 있을 경우 edge에서도 찾기
edge_terms = pd.concat(
    [edges["source_norm"], edges["target_norm"]],
    ignore_index=True
)

edge_seed_mask = regex_mask(
    edge_terms,
    STRICT_PATTERNS
)

seed_names.update(
    edge_terms.loc[edge_seed_mask].unique()
)

print("\nSeed node 목록:")
for name in sorted(seed_names):
    print(" ", name)


# ------------------------------------------------------------
# 8. 검색 범위별 엣지 수 비교
# ------------------------------------------------------------

strict_edge_mask = (
    edges["source_norm"].isin(seed_names)
    | edges["target_norm"].isin(seed_names)
)

generic_edge_mask = (
    regex_mask(edges["source_norm"], GENERIC_VIRUS_PATTERNS)
    | regex_mask(edges["target_norm"], GENERIC_VIRUS_PATTERNS)
)

immune_general_mask = (
    regex_mask(edges["source_norm"], IMMUNE_GENERAL_PATTERNS)
    | regex_mask(edges["target_norm"], IMMUNE_GENERAL_PATTERNS)
)

edges["candidate_class"] = np.select(
    [
        strict_edge_mask,
        generic_edge_mask,
        immune_general_mask
    ],
    [
        "SARS-CoV-2_direct",
        "generic_virus",
        "immune_general_only"
    ],
    default="not_virus_candidate"
)

print("\n=== 후보 분류별 엣지 수 ===")
print(
    edges["candidate_class"]
    .value_counts()
    .to_string()
)

print("\n=== 세포 KG별 SARS-CoV-2 직접 연결 엣지 수 ===")
print(
    edges.loc[strict_edge_mask]
    .groupby("cell_context")
    .size()
    .sort_values(ascending=False)
    .to_string()
)


# ------------------------------------------------------------
# 9. SARS-CoV-2 직접 연결 엣지만 추출
# ------------------------------------------------------------

covid_direct = edges.loc[strict_edge_mask].copy()

covid_direct = covid_direct.sort_values(
    ["weight", "cell_context"],
    ascending=[False, True]
)

print("\n=== SARS-CoV-2 직접 연결 엣지 상위 50개 ===")

display_columns = [
    "cell_context",
    "source",
    "relationship",
    "target",
    "weight",
    "file_count",
    "pmids"
]

print(
    covid_direct[display_columns]
    .head(50)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 10. activate / inhibit 분포
# ------------------------------------------------------------

print("\n=== 관계 방향 분포 ===")
print(
    pd.crosstab(
        covid_direct["cell_context"],
        covid_direct["relationship"]
    ).to_string()
)


# ------------------------------------------------------------
# 11. 동일한 노드 쌍에서 activate/inhibit 충돌 확인
# ------------------------------------------------------------

contradictions = (
    covid_direct
    .groupby(
        ["cell_context", "source_norm", "target_norm"]
    )["relationship"]
    .nunique()
    .reset_index(name="relationship_count")
)

contradictions = contradictions[
    contradictions["relationship_count"] > 1
]

print("\n=== activate / inhibit 충돌 노드 쌍 ===")
print(f"충돌 쌍 수: {len(contradictions):,}")

if len(contradictions) > 0:
    print(contradictions.head(30).to_string(index=False))


# ------------------------------------------------------------
# 12. weight와 실제 PMID 수 일치 여부
# ------------------------------------------------------------

weight_mismatch = covid_direct[
    covid_direct["weight"] != covid_direct["pmid_count_parsed"]
]

print("\n=== weight와 파싱된 PMID 수가 다른 엣지 ===")
print(f"{len(weight_mismatch):,}개")

if len(weight_mismatch) > 0:
    print(
        weight_mismatch[
            [
                "cell_context",
                "source",
                "target",
                "weight",
                "pmid_count_parsed",
                "sources"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 13. 결과 저장
# ------------------------------------------------------------

covid_direct.to_csv(
    "ICKG_SARS_CoV_2_direct_edges_audit.csv",
    index=False,
    encoding="utf-8-sig"
)

print(
    "\n저장 완료: "
    "ICKG_SARS_CoV_2_direct_edges_audit.csv"
)
```
