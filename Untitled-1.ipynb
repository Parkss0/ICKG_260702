```python
# ============================================
# Step 3-2: 공개 BioBERT NER 모델 로드 & 테스트
# ============================================
from transformers import pipeline

# d4data/biomedical-ner-all : 유전자, 질병, 세포 등
# 100+개 생의학 개체를 인식하는 공개 모델 (즉시 사용 가능)
print("모델 다운로드 중... (처음엔 1~2분 걸림)")

ner = pipeline(
    "token-classification",
    model="d4data/biomedical-ner-all",
    aggregation_strategy="simple"   # 쪼개진 토큰을 단어 단위로 합쳐줌
)

print("✅ 모델 로드 완료\n")

# ----- 테스트 문장들 (면역/바이러스 관련) -----
test_sentences = [
    "IL-15 activates GZMB expression in NK cells.",
    "The Omicron variant shows increased immune escape compared to Delta.",
    "TGF-beta suppresses IL-2 production in regulatory T cells.",
]

for i, sentence in enumerate(test_sentences, 1):
    print(f"\n{'='*55}")
    print(f"[{i}] {sentence}")
    print('='*55)

    entities = ner(sentence)

    if entities:
        for ent in entities:
            word = ent['word']
            label = ent['entity_group']
            score = ent['score']
            print(f"    {word:25s} → {label:20s} (신뢰도 {score:.2f})")
    else:
        print("    (개체 없음)")
```

```python
# ============================================
# Step 3-3: 유전자 특화 NER 모델 재테스트
# ============================================
from transformers import pipeline

test_sentences = [
    "IL-15 activates GZMB expression in NK cells.",
    "TGF-beta suppresses IL-2 production in regulatory T cells.",
]

# ----- 후보 1: 유전자 특화 BioBERT -----
print("="*55)
print("후보 1: alvaroalon2/biobert_genetic_ner")
print("="*55)
try:
    ner_gene = pipeline(
        "token-classification",
        model="alvaroalon2/biobert_genetic_ner",
        aggregation_strategy="simple"
    )
    for sentence in test_sentences:
        print(f"\n문장: {sentence}")
        for ent in ner_gene(sentence):
            print(f"    {ent['word']:25s} → {ent['entity_group']:15s} ({ent['score']:.2f})")
except Exception as e:
    print(f"에러: {e}")
```

```python
# ============================================
# Step 4-A: Ollama 파이썬 연결 테스트
# ============================================
import requests
import json

def ask_ollama(prompt, model="llama3.2:3b"):
    """Ollama에 프롬프트 보내고 답 받기"""
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": model,
            "prompt": prompt,
            "stream": False
        }
    )
    return response.json()["response"]

# 간단 테스트
print("테스트 중...")
answer = ask_ollama("Reply with exactly one word: Hello")
print(f"응답: {answer}")
print("✅ 연결 성공" if answer else "❌ 응답 없음")
```

```python
# ============================================
# Step 4-B: 관계 추출 (논문 프롬프트 방식)
# ============================================

def extract_relationship(abstract, entity1, entity2, model="llama3.2:3b"):
    """
    논문 방식: abstract 기준으로 entity1 → entity2 방향 관계 추출
    답은 Activate / Inhibit / NoRelation 중 하나
    """
    prompt = f"""You are a computational biologist reading a biomedical abstract to infer the relationship between two biomedical terms.

Abstract: {abstract}

Determine the directional relationship FROM "{entity1}" TO "{entity2}".
Choose exactly ONE:
- Activate : if {entity1} directly or indirectly promotes, increases, or activates {entity2}
- Inhibit  : if {entity1} directly or indirectly reduces, blocks, or suppresses {entity2}
- NoRelation : if there is no clear directional relationship

Rules:
- Base your answer ONLY on the abstract.
- Reply with ONLY the single word: Activate, Inhibit, or NoRelation. No explanation.

Answer:"""

    answer = ask_ollama(prompt, model).strip()
    # 답 정리 (첫 단어만)
    for label in ["Activate", "Inhibit", "NoRelation"]:
        if label.lower() in answer.lower():
            return label
    return "Unclear"


# ----- 테스트 abstract (아까 NER 테스트에 쓴 문장) -----
test_abstract = """
IL-15 is a critical cytokine for NK cell function. Studies show that
IL-15 activates GZMB and PRF1 expression, promoting cytotoxic activity.
IL-15 also suppresses IL-10 production in NK cells.
"""

# NER에서 뽑았다고 가정한 유전자들
genes = ["IL-15", "GZMB", "PRF1", "IL-10"]

# 유전자 쌍 만들기 (방향이 있으니 순서쌍)
import itertools
pairs = list(itertools.permutations(genes, 2))

print(f"총 {len(pairs)}개 쌍 분석 중...\n")
print("="*55)

edges = []  # 그래프 엣지 저장용
for e1, e2 in pairs:
    rel = extract_relationship(test_abstract, e1, e2)
    # 관계 있는 것만 출력 & 저장
    if rel in ["Activate", "Inhibit"]:
        print(f"  {e1:8s} --[{rel:8s}]--> {e2}")
        edges.append({"source": e1, "target": e2, "relationship": rel})

print("="*55)
print(f"\n✅ 관계 있는 엣지 {len(edges)}개 추출됨")
print("\n엣지 목록:")
for edge in edges:
    print(f"  {edge}")
```
