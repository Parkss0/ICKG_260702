```python
# ============================================
# Step 1-3: 전체 파일 트리 + 데이터 파일 탐색
# ============================================

import os

print("=== ICKG 전체 파일 트리 ===\n")
for root, dirs, files in os.walk("ICKG"):
    # .git 폴더는 건너뛰기
    if ".git" in root:
        continue
    level = root.replace("ICKG", "").count(os.sep)
    indent = "  " * level
    print(f"{indent}{os.path.basename(root)}/")
    for f in files:
        filepath = os.path.join(root, f)
        size = os.path.getsize(filepath)
        size_mb = size / (1024 * 1024)
        print(f"{indent}  {f}  ({size_mb:.2f} MB)")

# 그래프 데이터일 가능성이 있는 파일 찾기
print("\n\n=== 그래프/엣지 데이터 후보 파일 ===")
for root, dirs, files in os.walk("ICKG"):
    if ".git" in root:
        continue
    for f in files:
        if any(ext in f.lower() for ext in ['.csv', '.gml', '.gexf', '.graphml', '.json', '.pkl', 'edge', 'graph', 'node']):
            print(f"  → {os.path.join(root, f)}")
```

```python
# ============================================
# Step 2: NER config 파일 내용 확인
# ============================================
import json

configs = [
    "ICKG/finetuned_NER_models/config_BioNLP13PC.json",
    "ICKG/finetuned_NER_models/config_NCBI-disease.json",
    "ICKG/finetuned_NER_models/config_GENIAcelltype.json",
]

for cfg_path in configs:
    print(f"\n{'='*50}")
    print(f"파일: {cfg_path.split('/')[-1]}")
    print('='*50)
    with open(cfg_path) as f:
        cfg = json.load(f)
    # 핵심 정보만 출력
    print(f"  base model     : {cfg.get('_name_or_path', 'N/A')}")
    print(f"  architecture   : {cfg.get('architectures', 'N/A')}")
    print(f"  num_labels     : {cfg.get('num_labels', len(cfg.get('id2label', {})))}")
    print(f"  labels         : {list(cfg.get('id2label', {}).values())}")

# readme 파일도 확인
print(f"\n{'='*50}")
print("readme 내용")
print('='*50)
with open("ICKG/finetuned_NER_models/readme") as f:
    print(f.read())
```
